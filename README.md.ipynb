
# Grocery Store SQL Analysis using MySQL

## Project Overview

This project focuses on analyzing grocery store data using MySQL. A relational database was designed with seven interconnected tables to manage customers, products, suppliers, employees, orders, and categories. SQL queries were used to analyze sales performance and generate business insights.

## Objectives

- Analyze customer purchasing behavior
- Identify top-selling products
- Calculate product-wise and supplier-wise revenue
- Analyze monthly sales and order trends
- Evaluate employee performance
- Understand product and category performance

## Dataset

- **Source:** Grocery Store Management Dataset
- **Database:** MySQL
- **Tables:** 7 relational tables
- **Key Tables:**
  - Suppliers
  - Categories
  - Products
  - Customers
  - Employees
  - Orders
  - Order Details

## Tools Used

- MySQL
- SQL
- MySQL Workbench

## Analysis Performed

- Database and Table Creation
- Primary Key and Foreign Key Constraints
- Data Retrieval and Filtering
- INNER JOIN and LEFT JOIN
- Aggregate Functions (COUNT, SUM, AVG)
- GROUP BY and HAVING
- Subqueries
- Sorting and Ranking
- Date Functions and Monthly Sales Analysis
- Customer, Product, Supplier, and Employee Analysis

## Results

- Identified top customers based on purchase amounts
- Analyzed product-wise revenue
- Compared product counts across categories
- Examined monthly sales and order trends
- Evaluated supplier contributions to product availability and revenue
- Compared employee performance using order counts and sales value
